Karena semua edge memiliki bobot positif dan heuristik nol, maka jalur yang dihasilkan adalah jalur dengan total bobot minimum, misalnya dari A ke C (2), C ke F (3), F ke H (5) total 10, atau mungkin A ke C ke G ke H (2+4+3=9), atau A ke B ke E ke H (5+4+5=14), dan seterusnya. Program ini akan memilih yang paling kecil. Disini tidak ada penanganan error karena input sudah ditentukan di dalam kode, jadi tidak perlu try except. Tujuan dari kode ini adalah memperlihatkan cara membuat graf, menggambarnya, melakukan penelusuran DFS, dan mencari jalur terpendek dengan metode UCS menggunakan A* dengan heuristik nol.

```python
import networkx as nx
import heapq  # priority queue

# Fungsi UCS manual (biar kelihatan prosesnya)
def ucs(graph, start, goal):
    # priority queue: (cost, node, path)
    queue = [(0, start, [start])]
    visited = set()

    step = 1

    print("=== MULAI UNIFORM COST SEARCH ===\n")

    while queue:
        # Ambil node dengan cost terkecil
        cost, current, path = heapq.heappop(queue)

        print(f"--- Step {step} ---")
        print("Node diambil :", current)
        print("Cost sejauh ini :", cost)
        print("Path :", path)

        # Jika sudah sampai goal
        if current == goal:
            print("\n🎯 GOAL DITEMUKAN!")
            return path, cost

        # Jika belum dikunjungi
        if current not in visited:
            visited.add(current)

            print("Ekspansi tetangga:")

            for neighbor in graph.neighbors(current):
                weight = graph[current][neighbor]['weight']
                new_cost = cost + weight
                new_path = path + [neighbor]

                print(f"  -> {current} ke {neighbor} | cost = {new_cost}")

                heapq.heappush(queue, (new_cost, neighbor, new_path))

        print()
        step += 1

    return None, float('inf')


# =======================
# DATA GRAF
# =======================
G = nx.Graph()
edges = [
    ('A', 'B', 5), ('A', 'C', 2),
    ('B', 'D', 8), ('B', 'E', 4),
    ('C', 'F', 3), ('C', 'G', 4),
    ('D', 'H', 2), ('E', 'H', 5),
    ('F', 'H', 5), ('G', 'H', 3)
]

G.add_weighted_edges_from(edges)

# =======================
# JALANKAN UCS
# =======================
path, cost = ucs(G, 'A', 'H')

print("\n=== HASIL AKHIR ===")
print("Path terbaik :", path)
print("Total cost   :", cost)
```

# Penjelasan

Program ini berfungsi untuk mencari jalur terbaik dari node A menuju node H menggunakan algoritma Uniform Cost Search (UCS) yang memprioritaskan pengiriman berdasarkan cost terkecil. Karena itu, jalur yang dihasilkan bukan yang paling sedikit node-nya, tapi yang total bobotnya paling kecil. Pertama, kode mengimport dua modul, yaitu networkx untuk membuat dan memproses graf, lalu heapq untuk keperluan priority queue (antrean prioritas). Makanya, antrean ini otomatis akan mengurutkan data berdasarkan cost dari yang paling kecil ke paling besar.

Di dalam fungsi ucs, terdapat parameter graph, start, dan goal. Fungsi ini tidak menggunakan return berupa nilai saja, tapi mengembalikan dua hal yaitu path dan total cost, walaupun sebenarnya tetap disebut fungsi karena ada return. Kemudian di dalamnya, priority queue diberi nama queue yang diisi tuple berisi (cost, node, path). Awalnya queue berisi (0, start, [start]) karena dari start ke start costnya 0 dan path hanya node start itu sendiri. Lalu ada variabel visited bertipe set untuk mencatat node mana saja yang sudah pernah diekspansi, dan variabel step untuk menghitung langkah ke berapa saat proses running. Setelah itu, program mencetak "=== MULAI UNIFORM COST SEARCH ===" sebagai penanda bahwa proses pencarian dimulai.

Selanjutnya ada perulangan while selama queue tidak kosong. Di setiap iterasi, program mengambil tuple dengan cost terkecil menggunakan heapq.heappop. Isi tuple tersebut disimpan ke variabel cost, current, dan path. Kemudian program menampilkan step ke berapa, node yang sedang diambil, cost sejauh ini, dan path yang ditempuh. Jika current sama dengan goal, maka program mencetak "GOAL DITEMUKAN!" dan langsung mengembalikan path beserta cost-nya. Tapi jika belum sampai goal, program memeriksa apakah current sudah pernah dikunjungi. Jika belum, maka current ditambahkan ke visited. Setelah itu, program mencetak "Ekspansi tetangga:" lalu melakukan perulangan untuk setiap neighbor (tetangga) dari current. Untuk setiap neighbor, program mengambil bobot (weight) dari edge yang menghubungkan current dan neighbor. Bobot ini sudah ditentukan sebelumnya di data graf. Kemudian new_cost dihitung dari cost sekarang ditambah weight, dan new_path dibuat dengan menambahkan neighbor ke path lama. Program mencetak informasi dari current ke neighbor beserta new_cost-nya. Setelah itu, program memasukkan tuple (new_cost, neighbor, new_path) ke dalam priority queue menggunakan heapq.heappush. Proses ini berulang terus sampai queue kosong atau goal ditemukan. Jika queue kosong dan goal tidak pernah ditemukan, maka fungsi mengembalikan None dan float('inf') yang berarti cost tak terhingga.

Di luar fungsi, program membuat graf kosong dari networkx, kemudian mendefinisikan daftar edges berupa tuple (node1, node2, weight). Contohnya dari A ke B bobot 5, A ke C bobot 2, B ke D bobot 8, dan seterusnya sampai G ke H bobot 3. Semua edges ini ditambahkan ke graf menggunakan method add_weighted_edges_from. Setelah graf siap, program memanggil fungsi ucs dengan parameter G (graf), 'A' sebagai start, dan 'H' sebagai goal. Hasil dari pemanggilan fungsi ini ditampung ke variabel path dan cost. Terakhir, program mencetak "=== HASIL AKHIR ===" lalu menampilkan path terbaik dan total cost. Sebenarnya program ini tidak terpaku hanya untuk node A ke H, bisa saja diganti start dan goal-nya ke node lain, tinggal mengubah parameter saat memanggil fungsi ucs. Tapi karena di soal atau keperluan ini targetnya dari A ke H, maka jalur terbaik yang ditemukan adalah A -> C (cost 2), C -> G (cost 4), G -> H (cost 3), sehingga total cost = 9 dengan path ['A', 'C', 'G', 'H']. Makanya, walaupun ada jalur lain seperti A -> B -> E -> H dengan total cost 5+4+5=14 atau A -> C -> F -> H dengan total 2+3+5=10, tetap jalur yang dipilih adalah yang paling kecil cost-nya, yaitu 9. Itu terjadi karena algoritma UCS selalu mengeksplorasi node dengan cost terkecil terlebih dahulu, sehingga ketika goal ditemukan, tidak ada kemungkinan jalur lain yang lebih murah.
