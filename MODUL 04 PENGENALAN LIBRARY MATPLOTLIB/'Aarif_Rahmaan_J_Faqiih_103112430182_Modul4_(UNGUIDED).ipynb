### 2. Membuat Plot Pairwise pada Matplotlib (25)

**a. (8 poin)** Sebuah toko sedang memantau penjualan dua produk unggulan mereka, **Barang A** dan **Barang B**, selama lima hari terakhir. Manajer penjualan ingin mengetahui pola penjualan dari kedua produk ini dengan membandingkan kinjerja keduanya dari hari ke hari. Buat visualisasi data penjualan dengan menggunakan *line chart* untuk mempermudah analisis tren.

Bantu manajer penjualan dengan membuat grafik *line chart* berdasarkan penjualan berikut:
- Menampilkan tren penjualan kedua **Barang A** dan **Barang B** dalam satu graf yang sama.
- Ukuran *figure* sebesar 7x5 dengan reolusi 100 dpi.
- Beri judul grafik dengan : `"Tren Penjualan Barang A dan B Selama 5 Hari"`.
- Sumbu $x$ mewakili hari ke berapa, dengan label `"Tanggal"`.
- Sumbu $y$ mewakili jumlah barang yang terjual, dengan label `"Jumlah Terjual"`.
- Tampilkan *legend* di pojok kiri bawah untuk membedakan penjualan Barang A dan Barang B.

Data Penjualan:
|Hari|Barang A|Barang B|
|-------|--------|--------|
|1|10|8|
|2|7|8|
|3|8|5|
|4|12|7|
|5|6|11|

Tip :

Gunakan `plt.figure()`

````python
# Inisialisasi figure dengan ketentuan yang sesuai
fig = plt.figure(figsize=(7, 5), dpi=100)

# List data penjualan
hari = [1, 2, 3, 4, 5]
barang_a = [10, 7, 8, 12, 6]
barang_b = [8, 8, 5, 7, 11]

# Plot linechart Barang A pada figure
plt.plot(hari, barang_a, label="Barang A")

# Plot linechart Barang B pada figure
plt.plot(hari, barang_b, label="Barang B")

# Menambahkan judul, dan label untuk x dan y
plt.title("Tren Penjualan Barang A dan B Selama 5 Hari")
plt.xlabel("Tanggal")
plt.ylabel("Jumlah Terjual")

# Menambahkan legend dengan lokasi yang ditentukan
plt.legend(loc='lower left')

# Tampilkan linechart yang sudah dibuat
plt.show()
````

**Penjelasan**

program ini berfungsi untuk menampilkan visualisasi tren penjualan dua barang yaitu barang a dan barang b selama 5 hari dalam bentuk line chart, pertama dilakukan inisialisasi figure menggunakan plt.figure dengan ukuran dan dpi yang sudah ditentukan, kemudian disiapkan data berupa list yang berisi hari sebagai sumbu x dan jumlah penjualan untuk masing-masing barang yaitu barang_a dan barang_b sebagai sumbu y.

selanjutnya program membuat line chart untuk barang a menggunakan plt.plot dengan label yang sesuai, kemudian dibuat juga line chart untuk barang b dengan cara yang sama agar kedua data bisa dibandingkan dalam satu grafik.

setelah itu program menambahkan judul dan label untuk sumbu x dan y menggunakan plt.title, plt.xlabel, dan plt.ylabel, kemudian ditambahkan legend menggunakan plt.legend dengan posisi di kiri bawah agar user bisa mengetahui garis mana untuk barang a dan barang b, dan terakhir ditampilkan menggunakan plt.show(), jadi user bisa melihat perbandingan tren penjualan kedua barang selama 5 hari.

- Sumbu $x$ menunjukkan minggu ke berapa, dengan label `"Minggu"`
- Sumbu $y$ menunjukkan tinggi tanaman dalam cm, dengan label `"Tinggi Tanaman (cm)"`
- Tampilkan *legend* untuk membedakan titik data dari kedua tanaman tersebut.

Data pertumbuhan tanaman
|Minggu|Tinggi Tanaman A (cm)|Tinggi Tanaman B (cm)|
|---|---|---|

|2|21|20|
|3|25|26|
|4|30|29|

Gunakan atribut berikut pada *figure* visualisasi yang dibuat.

- Ukuran *figure* adalah 8x6 dengan 75 dpi.
- Judul grafik yang dibuat: `"Penjualan Buku Berdasarkan Kategori Selama Sebulan"`
- Label sumbu $x$: `"Kategori Buku"`
- Label sumbu $y$: `"Jumlah Terjual"`
- Gunakan warna yang berbeda untuk setiap bar

Data penjualan:
|Kategori Buku|Jumlah Terjual|
|---|---|

|Non-Fiksi|90|
|Komik|150|
|Majalah|80|

Tip:

Gunakan `plt.subplots()`

````python
# List data penjualan buku
kategori = ['Fiksi', 'Non-Fiksi', 'Komik', 'Majalah', 'Sains']
jumlah_terjual = [120, 90, 150, 80, 110]

# Inisialisasi figure dengan ukuran dan dpi yang ditentukan
fig, ax = plt.subplots(figsize=(8, 6), dpi=75)

# List warna pada barchart
list_color = ['red', 'blue', 'green', 'purple', 'orange']
# Plot barchart untuk data penjualan buku
ax.bar(kategori, jumlah_terjual, color=list_color)

# Menambahkan judul dan label untuk x dan y
ax.set_title("Penjualan Buku Berdasarkan Kategori Selama Sebulan")
ax.set_xlabel("Kategori Buku")
ax.set_ylabel("Jumlah Terjual")

# Tampilkan barchart yang sudah dibuat
plt.show()
````

### 3. Membuat Plot Distribusi Statistik pada Matplotlib (20)

**a. (6 poin)** Seorang peneliti ingin menganalisis distribusi waktu belajar harian (dalam jam) dari sekelompok mahasiswa selama satu minggu. Peneliti telah mengumpulkan data dari 50 mahasiswa, di mana setiap mahasiswa mencatat waktu belajar mereka setiap hari selama seminggu. Berdasarkan data yang dikumpulkan, peneliti ingin mengetahui distribusi waktu belajar tersebut dalam bentuk histogram.

Buatlah histogram untuk membantu peneliti memvisualisasikan distribusi waktu belajar mahasiswa. Gunakan atribut berikut:

- Ukuran *figure* adalah 10x5 dengan resolusi 100dpi.
- Judul grafik: `"Distribusi Waktu Belajar Mahasiswa Selama Seminggu"`.
- Label sumbu $x$: `"Waktu Belajar (Jam)"`.
- Label sumbu $y$: `"Jumlah Mahasiswa"`.
- Karena nilai minuimum pada data adalah 1 dan nilai maksimum adalah 7, maka jumlah *bins* menjadi 7 (1 bin merepresentasikan 1 jam).

Berikut list data dari 50 mahasiswa
```python
[2, 3, 5, 4, 6, 2, 1, 3, 2, 4, 5, 2, 3, 1, 6, 3, 5, 4, 4, 2, 2, 3, 5, 1, 3, 5, 6, 7, 4, 5, 2, 3, 4, 1, 2, 3, 4, 6, 2, 5, 4, 3, 5, 2, 3, 6, 7, 5, 4, 3]
```

Tip:

Gunakan `plt.subplots()`

````python
# Data waktu belajar harian mahasiswa
data_waktu_belajar = [2, 3, 5, 4, 6, 2, 1, 3, 2, 4, 5, 2, 3, 1, 6, 3, 5, 4, 4, 2,
                      2, 3, 5, 1, 3, 5, 6, 7, 4, 5, 2, 3, 4, 1, 2, 3, 4, 6, 2, 5,
                      4, 3, 5, 2, 3, 6, 7, 5, 4, 3]

# Inisialisasi figure dengan ukuran dan dpi yang ditentukan
fig, ax = plt.subplots(figsize=(10, 5), dpi=100)

# Plot histogram waktu belajar
ax.hist(data_waktu_belajar, bins=7, color='teal', edgecolor='black')

# Menambahkan judul dan label untuk x dan y
ax.set_title("Distribusi Waktu Belajar Mahasiswa Selama Seminggu")
ax.set_xlabel("Waktu Belajar (Jam)")
ax.set_ylabel("Jumlah Mahasiswa")

# Tampilkan histogram yang sudah dibuat
plt.show()
````

**Penjelasan**

program ini berfungsi untuk menampilkan visualisasi distribusi waktu belajar harian mahasiswa dalam bentuk histogram, pertama disiapkan data berupa list yang berisi jam belajar mahasiswa dalam sehari yang disimpan dalam variabel data_waktu_belajar, data ini nantinya akan digunakan untuk melihat persebaran kebiasaan belajar mahasiswa.

selanjutnya program melakukan inisialisasi figure menggunakan plt.subplots dengan ukuran dan dpi yang sudah ditentukan, kemudian digunakan ax.hist untuk membuat histogram dari data_waktu_belajar dengan jumlah bins tertentu untuk mengelompokkan data, serta ditambahkan warna dan garis tepi agar tampilannya lebih jelas.

setelah itu program menambahkan judul dan label untuk sumbu x dan y menggunakan set_title, set_xlabel, dan set_ylabel, sehingga grafik menjadi lebih mudah dipahami, dan terakhir ditampilkan menggunakan plt.show(), jadi user bisa melihat distribusi waktu belajar mahasiswa selama seminggu.

**b. (7 poin)** Sebuah perusahaan *e-commerce* ingin menganalisis distribusi kategori produk yang pailng sering dibeli oleh pelanggan selama bulan terakhir. Data yang dikumpulkan menunjukkan bahwa terdapat lima kategori porduk yang paling sering dibeli: **Elektronik**, **Pakaian**, **Peralatan Rumah Tangga**, **Buku**, dan **Kosmetik**. Perusahaan ingin melihat presentase kontribusi dari setiap kategori produk dalam bentuk *pie chart*.

Bantu tim analisis perusahaan dengan membuat *pie chart* untuk memvisualisasikan distribusi pembelian produk berdasarkan kategori berikut.

- **Elektronik**: 1500 pembelian.
- **Pakaian**: 1200 pembelian.
- **Peralatan Rumah Tangga**: 900 pembelian.
- **Buku**: 800 pembelian.
- **Komestik**: 600 pembelian.

Gunakan atribut berikut:
- Ukuran *figure* adalah 8x8 dengan resolusi 100 dpi.
- Judul grafik: `"Distribusi Pembelian Produk Berdasarkan Kategori"`.
- Tambahkan persentase pada setiap bagian *pie chart*.
- Pisahkan potongan *pie chart* untuk kategori dengan jumlah pembelian tertinggi.
- Gunakan warna yang berbeda untuk setiap kategori.

````python
# List data kategori produk dan jumlah pembelian
kategori = ['Elektronik', 'Pakaian', 'Peralatan Rumah Tangga', 'Buku', 'Kosmetik']
jumlah_pembelian = [1500, 1200, 900, 800, 600]

# Inisialisasi figure dengan ukuran dan resolusi yang sudah ditentukan
fig, ax = plt.subplots(figsize=(8, 8), dpi=100)

# Membuat explode untuk memisahkan kategori dengan pembelian tertinggi
explode = [0.1, 0, 0, 0, 0]

# List warna pada piechart
color_list = ['skyblue', 'lightgreen', 'lightcoral', 'orange', 'purple']

# Plot piechart untuk data penjualan produk
ax.pie(jumlah_pembelian, labels=kategori, autopct='%1.1f%%', explode=explode, colors=color_list)

# Menambahkan judul
ax.set_title("Distribusi Pembelian Produk Berdasarkan Kategori")

# Tampilan piechart yang sudah dibuat
plt.show()
````

**Penjelasan**

program ini berfungsi untuk menampilkan visualisasi distribusi pembelian produk berdasarkan kategori dalam bentuk pie chart, pertama disiapkan data berupa list yang berisi kategori produk dan jumlah pembelian dari masing-masing kategori seperti elektronik, pakaian, peralatan rumah tangga, buku, dan kosmetik, data tersebut disimpan dalam variabel kategori dan jumlah_pembelian.

selanjutnya program melakukan inisialisasi figure menggunakan plt.subplots dengan ukuran dan resolusi yang sudah ditentukan, kemudian dibuat variabel explode untuk memisahkan kategori dengan pembelian tertinggi agar lebih menonjol, di sini nilai 0.1 diberikan pada elemen pertama karena itu adalah kategori dengan jumlah pembelian terbesar.

kemudian dibuat list warna yaitu color_list yang berisi beberapa warna untuk mempercantik tampilan pie chart, setelah itu digunakan ax.pie untuk membuat pie chart dengan parameter labels untuk nama kategori, autopct untuk menampilkan persentase, explode untuk memberikan efek pemisahan, dan colors untuk memberikan warna pada setiap bagian.

setelah itu program menambahkan judul menggunakan set_title agar grafik lebih jelas, dan terakhir ditampilkan menggunakan plt.show(), jadi user bisa melihat perbandingan pembelian produk berdasarkan kategori dalam bentuk persentase.

**c. (7 poin)** Sebuah sekolah sedang menganalisis nilai ujian akhir dari tiga kelas yang berbeda: **Kelas A**, **Kelas B**, **Kelas C**. Setiap kelas diikuti oleh sejumlah siswa, dan mereka telah menyelesaikan ujian akhir dalam mata pelajaran Matematika. Sekolah inigin memvisualisasikan sebaran nilai ujian dari masing-masing kelas untuk membandingkan performa antar kelas tersebut.

Bantu pihak sekolah dengan membuat *boxplot* yang menampilkan distribusi nilai ujian dari ketiga kelas tersebut. Data nilai ujian dari masing-masing kelas adalah sebagai berikut:
- **Kelas A**: [85, 78, 92, 88, 76, 90, 85, 91, 89, 95]
- **Kelas B**: [72, 75, 78, 74, 73, 80, 76, 77, 79, 81]
- **Kelas C**: [88, 90, 85, 93, 87, 92, 91, 86, 84, 94]

Gunakan atribut berikut dalam visualisasi:
- Ukuran *figure* adalah 8x6 dengan resolusi 100 dpi.
- Judul grafik: `"Distribusi Nilai Ujian Akhir Matematika di Kelas A, B, dan C"`.
- Label sumbu $x$: `"Kelas"`.
- Label sumbu $y$: `"Nilai Ujian"`.
- Gunakan warna yang berbeda untuk setiap *boxplot*.

````python
# List data nilai pada ketiga kelas
nilai_kelas_a = [85, 78, 92, 88, 76, 90, 85, 91, 89, 95]
nilai_kelas_b = [72, 75, 78, 74, 73, 80, 76, 77, 79, 81]
nilai_kelas_c = [88, 90, 85, 93, 87, 92, 91, 86, 84, 94]

# Menyiapkan data untuk boxplot
data = [nilai_kelas_a, nilai_kelas_b, nilai_kelas_c]
labels = ['Kelas A', 'Kelas B', 'Kelas C']

# Inisialisasi figure dengan ukuran dan dpi yang sudah ditentukan
fig, ax = plt.subplots(figsize=(8, 6), dpi=100)

# Plot boxplot untuk data nilai ujian
# Note: 'labels' parameter is renamed to 'tick_labels' in newer Matplotlib versions
bp = ax.boxplot(data, vert=True, tick_labels=labels, patch_artist=True)

# Menambahkan warna yang berbeda untuk setiap boxplot
colors = ['cyan', 'lightgreen', 'pink']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

# Menambahkan judul dan label untuk x dan y
ax.set_title("Distribusi Nilai Ujian Akhir Matematika di Kelas A, B, dan C")
ax.set_xlabel("Kelas")
ax.set_ylabel("Nilai Ujian")

# Menampilkan plot
plt.show()
````

**Penjelasan**

program ini berfungsi untuk menampilkan visualisasi distribusi nilai ujian akhir matematika dari tiga kelas yaitu kelas a, b, dan c dalam bentuk boxplot, pertama disiapkan data berupa list yang berisi nilai dari masing-masing kelas yaitu nilai_kelas_a, nilai_kelas_b, dan nilai_kelas_c, kemudian semua data tersebut digabungkan ke dalam satu variabel yaitu data, dan label untuk masing-masing kelas disimpan dalam variabel labels.

Sebuah universitas sedang melakukan analisis performa akademik mahasiswa di empat jurusan: **Informatika**, **Teknologi Informasi**, **Rekayasa Perangkat Lunak**, dan **Sains Data**. Universitas ingin menganalisis beberapa aspek berikut :

**1. (10 poin) Distribusi Nilai Akhir Mahasiswa** <br>
Setiap jurusan mengumpulkan nilai akhir mahasiswa dalam tiga mata kuliah inti selama satu semester. Universitas ingin melihat distribusi nilai dari masing-masing jurusan.
    - **Informatika**: [85, 78, 92, 88, 76, 90, 85, 91, 89, 95].
    - **Teknologi Informasi**: [72, 75, 78, 74, 73, 80, 76, 77, 79, 81].
    - **Rekayasa Perangkat Lunak**: [88, 90, 85, 93, 87, 92, 91, 86, 84, 94].
    - **Sains Data**: [80, 82, 78, 85, 84, 83, 82, 88, 86, 87].

    Buatlah *boxplot* untuk memvisualisasikan distribusi nilai akhir dari keempat jurusan tersebut.

**2. (10 poin) Rata-rata Kehadiran Mahasiswa** <br>
Universitas juga ingin mengetahui rata-rata kehadiran mahasiswa per jurusan selama semester ini. Data kehadiran dalam persen adalah sebagai berikut:
    - **Informatika**: $85\%$.
    - **Teknologi Informasi**: $90\%$.
    - **Rekayasa Perangkat Lunak**: $88\%$.

    Buatlah *barplot* untuk memvisualisasikan rata-rata kehadiran mahasiswa per jurusan.

**3. (10 poin) Proporsi Mahasiswa Berdasarkan Jurusan** <br>
Universitas ingin melihat proporsi jumlah mahasiswa di setiap jurusan untuk memahami distribusi populasi. Data jumlah mahasiswa per jurusan adalah sebagai berikut:
    - **Informatika**: 300 Mahasiswa.
    - **Teknologi Informasi**: 250 Mahasiswa.
    - **Rekayasa Perangkat Lunak**: 200 Mahasiswa.
    - **Sains Data**: 150 Mahasiswa.

**4. (10 poin) Distribusi Jam Belajar Mahasiswa** <br>
Universitas juga melakukan survei terkait jam belajar harian dari 50 mahasiswa secara acak. Data waktu belajar harian (dalam jam) yang diukur dari mahasiswa adalah sebagai berikut

    ```python
    [2, 3, 5, 4, 6, 2, 1, 3, 2, 4, 5, 2, 3, 1, 6, 3, 5, 4, 4, 2, 2, 3, 5, 1, 3, 5, 6, 7, 4, 5, 2, 3, 4, 1, 2, 3, 4, 6, 2, 5, 4, 3, 5, 2, 3, 6, 7, 5, 4, 3]
    ```

    Buatlah histogram untuk menampilkan distribusi jam belajar harian dari mahasiswa.

Gunakan atribut berikut pada setiap visualisasi:
- Ukuran *figure* untuk semua grafik adalah 8x6 dengan resolusi 100 dpi.
- Beri judul yang sesuai untuk setiap grafik.
- Berikan label sumbu pada *barplot*, *boxplot*, dan *histogram*.
- Tambahkan presentase pada *pie chart*.

````python
# 1. Data nilai akhir mahasiswa
nilai_informatika = [85, 78, 92, 88, 76, 90, 85, 91, 89, 95]
nilai_teknologi_informasi = [72, 75, 78, 74, 73, 80, 76, 77, 79, 81]
nilai_rekayasa_perangkat_lunak = [88, 90, 85, 93, 87, 92, 91, 86, 84, 94]
nilai_sains_data = [80, 82, 78, 85, 84, 83, 82, 88, 86, 87]

fig1, ax1 = plt.subplots(figsize=(8, 6), dpi=100)
data_nilai = [nilai_informatika, nilai_teknologi_informasi, nilai_rekayasa_perangkat_lunak, nilai_sains_data]
labels_jurusan = ['Informatika', 'Teknologi Informasi', 'RPL', 'Sains Data']
ax1.boxplot(data_nilai, tick_labels=labels_jurusan, patch_artist=True)
ax1.set_title("Distribusi Nilai Akhir Mahasiswa per Jurusan")
ax1.set_xlabel("Jurusan")
ax1.set_ylabel("Nilai Akhir")
plt.show()

# 2. Data rata-rata kehadiran mahasiswa
jurusan = ['Informatika', 'Teknologi Informasi', 'RPL', 'Sains Data']
kehadiran = [85, 90, 88, 87]

fig2, ax2 = plt.subplots(figsize=(8, 6), dpi=100)
ax2.bar(jurusan, kehadiran, color=['teal', 'orange', 'green', 'red'])
ax2.set_title("Rata-rata Kehadiran Mahasiswa per Jurusan")
ax2.set_xlabel("Jurusan")
ax2.set_ylabel("Kehadiran (%)")
plt.show()

# 3. Data jumlah mahasiswa berdasarkan jurusan
jurusan = ['Informatika', 'Teknologi Informasi', 'RPL', 'Sains Data']
mahasiswa = [300, 250, 200, 150]

fig3, ax3 = plt.subplots(figsize=(8, 6), dpi=100)
ax3.pie(mahasiswa, labels=jurusan, autopct='%1.1f%%', startangle=140)
ax3.set_title("Proporsi Mahasiswa Berdasarkan Jurusan")
plt.show()

# 4. Data jam belajar harian mahasiswa
data_waktu_belajar = [2, 3, 5, 4, 6, 2, 1, 3, 2, 4, 5, 2, 3, 1, 6, 3, 5, 4, 4, 2,
                      2, 3, 5, 1, 3, 5, 6, 7, 4, 5, 2, 3, 4, 1, 2, 3, 4, 6, 2, 5,
                      4, 3, 5, 2, 3, 6, 7, 5, 4, 3]

fig4, ax4 = plt.subplots(figsize=(8, 6), dpi=100)
ax4.hist(data_waktu_belajar, bins=7, color='skyblue', edgecolor='black')
ax4.set_title("Distribusi Jam Belajar Harian Mahasiswa")
ax4.set_xlabel("Jam Belajar")
ax4.set_ylabel("Jumlah Mahasiswa")
plt.show()
````

**Penjelasan**

program ini berfungsi untuk menampilkan berbagai visualisasi data mahasiswa dalam bentuk grafik, mulai dari distribusi nilai, kehadiran, jumlah mahasiswa, sampai jam belajar harian, semua dijadikan dalam satu file dengan beberapa bagian grafik yang berbeda. pertama disiapkan data berupa list yang berisi nilai akhir mahasiswa dari beberapa jurusan seperti informatika, teknologi informasi, rekayasa perangkat lunak, dan sains data, data-data ini nantinya akan digunakan untuk divisualisasikan.

di bagian pertama, program membuat boxplot untuk menampilkan distribusi nilai akhir mahasiswa per jurusan, data nilai dari setiap jurusan dimasukkan ke dalam satu variabel yaitu data_nilai, kemudian label jurusan disimpan di labels_jurusan, setelah itu digunakan ax1.boxplot untuk membuat boxplot dengan label yang sesuai, lalu ditambahkan judul, label sumbu x dan y menggunakan set_title, set_xlabel, dan set_ylabel, kemudian ditampilkan dengan plt.show(), jadi user bisa melihat persebaran nilai dari masing-masing jurusan.

selanjutnya program menampilkan grafik batang untuk rata-rata kehadiran mahasiswa, data jurusan dan kehadiran disimpan dalam list, kemudian dibuat bar chart menggunakan ax2.bar dengan tambahan warna agar lebih jelas, setelah itu diberi judul dan label sumbu, lalu ditampilkan menggunakan plt.show(), sehingga user bisa melihat perbandingan kehadiran antar jurusan.

kemudian program membuat pie chart untuk menunjukkan proporsi jumlah mahasiswa berdasarkan jurusan, data jumlah mahasiswa disimpan dalam list mahasiswa dan label jurusan disimpan di variabel jurusan, lalu digunakan ax3.pie untuk membuat diagram lingkaran dengan persentase menggunakan autopct dan pengaturan sudut awal dengan startangle, setelah itu diberi judul dan ditampilkan dengan plt.show(), jadi bisa terlihat perbandingan jumlah mahasiswa dalam bentuk persentase.

terakhir program menampilkan histogram untuk distribusi jam belajar harian mahasiswa, data jam belajar disimpan dalam list data_waktu_belajar, kemudian dibuat histogram menggunakan ax4.hist dengan jumlah bins tertentu untuk mengelompokkan data, serta ditambahkan warna dan garis tepi agar lebih jelas, setelah itu diberi judul dan label sumbu, lalu ditampilkan dengan plt.show(), sehingga user bisa melihat distribusi kebiasaan belajar mahasiswa dalam sehari.
