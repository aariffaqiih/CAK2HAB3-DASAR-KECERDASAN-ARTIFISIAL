program ini berfungsi untuk membuat line chart menggunakan matplotlib untuk menampilkan dua data dalam bentuk garis, pertama kita siapkan data berupa x sebagai sumbu x dan dua data yaitu y1 dan y2 yang akan ditampilkan. setelah itu kita buat figure dengan plt.figure(figsize=(10,6)) untuk mengatur ukuran tampilan plot. lalu membuat line chart menggunakan plt.plot() dengan memasukkan x dan y1 sebagai data pertama dan x dan y2 sebagai data kedua serta label untuk masing-masing data. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditambahkan legend dan ditampilkan hasil plotnya dengan plt.legend() dan plt.show()

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y1 = [10, 20, 25, 30]
y2 = [12, 18, 30, 40]

# Membuat figure dengan ukuran 8x6
fig = plt.figure(figsize=(8,6))

# Membuat linechart
plt.plot(x, y1, color="pink", linewidth=3, label="Data Y1")  # Garis hijau dengan ketebalan 3
plt.plot(x, y2, color = "blue", linestyle='--', label="Data Y2")  # Garis putus-putus

# Labeling plot
plt.title("Contoh Plot Garis")  # Judul grafik
plt.xlabel("Sumbu X")  # Label sumbu X
plt.ylabel("Sumbu Y")  # Label sumbu Y

# Menampilkan legenda
plt.legend()

# Menampilkan figure
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat line chart menggunakan matplotlib untuk menampilkan dua data dalam bentuk garis, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa x sebagai sumbu x dan dua data yaitu y1 dan y2 yang akan ditampilkan. setelah itu kita buat figure dengan plt.figure(figsize=(8,6)) untuk mengatur ukuran tampilan plot. lalu membuat line chart menggunakan plt.plot() dengan memasukkan x dan y1 sebagai data pertama dengan warna pink dan ketebalan garis, kemudian ditambahkan data kedua x dan y2 dengan warna biru serta linestyle putus-putus dan label masing-masing. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan legend dan juga hasil plotnya dengan plt.legend() dan plt.show()

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5, 6, 7, 8]
y1 = [10, 21, 25, 30, 34, 41, 47, 50]
y2 = [12, 20, 26, 29, 33, 37, 41, 43]

# Membuat figure dengan ukuran 10x5 inch
fig = plt.figure(figsize=(10,5))

# Membuat scatter plot
plt.scatter(x, y1, s=100, c='pink', alpha=0.65, label='Data 1')  # Titik merah, ukuran besar
plt.scatter(x, y2, s=50, c='green', alpha=0.65, label='Data 2')  # Titik biru, ukuran kecil

# Labeling plot
plt.title("Contoh Scatter Plot")  # Menambahkan judul
plt.xlabel("Data X")  # Label untuk sumbu X
plt.ylabel("Data Y")  # Label untuk sumbu Y

# Menampilkan legenda berdasarkan label pada data
plt.legend()

# Menampilkan figure
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat scatter plot menggunakan matplotlib untuk menampilkan sebaran dua data dalam bentuk titik, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa x sebagai sumbu x dan dua data yaitu y1 dan y2 yang akan ditampilkan. setelah itu kita buat figure dengan plt.figure(figsize=(10,5)) untuk mengatur ukuran tampilan plot. lalu membuat scatter plot menggunakan plt.scatter() dengan memasukkan x dan y1 sebagai data pertama dengan ukuran titik besar, warna pink, dan transparansi, kemudian ditambahkan data kedua x dan y2 dengan ukuran lebih kecil, warna hijau, dan label masing-masing. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan legend dan juga hasil plotnya dengan plt.legend() dan plt.show()

```python
# Data
kategori = ['A', 'B', 'C', 'D'] # Label setiap kategori
nilai = [10, 20, 25, 30] # Nilai setiap kategori
# Membuat Figure
fig = plt.figure(figsize=(6,8))
# Membuat bar plot dengan plt.bar()
plt.bar(kategori, nilai, hatch='/', color='skyblue')
# Labeling plot
plt.title("Contoh Bar Plot")
plt.xlabel("Kategori")
plt.ylabel("Nilai")
# Menampilkan figure
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat bar plot menggunakan matplotlib untuk menampilkan data dalam bentuk diagram batang, pertama kita siapkan data berupa kategori sebagai label dan nilai sebagai data yang akan ditampilkan. setelah itu kita buat figure dengan plt.figure(figsize=(6,8)) untuk mengatur ukuran tampilan plot. lalu membuat bar plot menggunakan plt.bar() dengan memasukkan kategori sebagai sumbu x dan nilai sebagai data, serta hatch='/' untuk memberikan pola garis dan color='skyblue' untuk memberi warna. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan hasil plotnya dengan plt.show()

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y1 = [9, 20, 27, 35]
y_main = [7, 18, 22, 29]  # Data garis utama
y2 = [6, 15, 18, 23]

# Membuat figure
plt.figure(figsize=(10, 6))

# Membuat fill_between plot
plt.fill_between(x, y1, y2, color='pink', alpha=0.5, label='Fill Plot')

# Membuat plot garis utama
plt.plot(x, y_main, color='red', linewidth=2, label='Garis Utama')

# Labeling plot
plt.title("Fill Between")
plt.xlabel("X")
plt.ylabel("Y")

# Menampilkan legend dan figure
plt.legend(loc='lower right') # legenda di kanan bawah
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat fill between plot menggunakan matplotlib untuk menampilkan area di antara dua data serta garis utama, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa x sebagai sumbu x dan beberapa data yaitu y1 dan y2 untuk area, serta y_main untuk garis utama. setelah itu kita buat figure dengan plt.figure(figsize=(10, 6)) untuk mengatur ukuran tampilan plot. lalu membuat fill between menggunakan plt.fill_between() dengan memasukkan x, y1, dan y2 sebagai batas area, serta color='pink' untuk warna dan alpha=0.5 untuk transparansi, lalu label untuk keterangan. kemudian kita buat garis utama menggunakan plt.plot() dengan data x dan y_main serta tambahan warna merah dan linewidth untuk ketebalan garis. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan legend dan juga hasil plotnya dengan plt.legend() dan plt.show()

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y1 = [10, 20, 25, 30]
y2 = [5, 10, 15, 20]
y3 = [2, 5, 8, 12]

# Membuat figure
fig = plt.figure(figsize=(10,6))

# Membuat stack plot
plt.stackplot(x, y1, y2, y3, labels=['Y1', 'Y2', 'Y3'])

# Labeling plot
plt.title("Stack Plot")
plt.xlabel("X")
plt.ylabel("Y")

# Menampilkan legend dan figure
plt.legend()
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat stack plot menggunakan matplotlib untuk menampilkan beberapa data yang ditumpuk dalam satu grafik, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa x sebagai sumbu x dan beberapa data y yaitu y1, y2, dan y3 yang akan ditampilkan. setelah itu kita buat figure dengan plt.figure(figsize=(10,6)) untuk mengatur ukuran tampilan plot. lalu membuat stack plot menggunakan plt.stackplot() dengan memasukkan x sebagai sumbu x dan y1, y2, y3 sebagai data, serta labels untuk memberi nama tiap data. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan legend dan juga hasil plotnya dengan plt.legend() dan plt.show()

```python
import matplotlib.pyplot as plt

# Data
data = [1,9,2,2,8,8,7,7,7,3,3,3,4,4,4,4,6,6,6,6,5,5,5,5,5,5]

# Membuat plot
fig = plt.figure(figsize=(10,6))

# Membuat histogram
plt.hist(data, bins=9, color="blue", alpha=0.7) # 9 bin (interval), warna biru, transparansi 0.7

# Labeling plot
plt.title("Contoh Histogram")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")

# Menampilkan figure
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat histogram menggunakan matplotlib untuk menampilkan frekuensi dari data, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa list angka yang akan divisualisasikan. setelah itu kita buat figure dengan plt.figure(figsize=(10,6)) untuk mengatur ukuran tampilan plot. lalu membuat histogram menggunakan plt.hist() dengan memasukkan data, bins=9 untuk menentukan jumlah interval, color="blue" untuk memberi warna biru, dan alpha=0.7 untuk mengatur transparansi. kemudian diberi judul menggunakan plt.title dan label sumbu x serta y menggunakan plt.xlabel dan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan hasil plotnya dengan plt.show()

```python
import matplotlib.pyplot as plt

# Data
data = [7, 5, 8, 9, 6, 7, 5, 8, 6, 7, 6, 8, 9, 4, 7, 5, 9, 8, 6, 7]

# Membuat plot
fig = plt.figure(figsize=(8,6))

# Membuat box plot
plt.boxplot(data, patch_artist=True, widths=0.50) #Lebar box = 0.50

# Labeling Plot
plt.title("Box Plot")
plt.ylabel("Nilai")

# Menampilkan figure
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat box plot menggunakan matplotlib untuk menampilkan distribusi data, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa list angka yang akan divisualisasikan. setelah itu kita buat figure dengan plt.figure(figsize=(8,6)) untuk mengatur ukuran tampilan plot. lalu membuat box plot menggunakan plt.boxplot() dengan memasukkan data, serta patch_artist=True agar box memiliki warna dan widths=0.50 untuk mengatur lebar box. kemudian diberi judul menggunakan plt.title dan label sumbu y menggunakan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan hasil plotnya dengan plt.show()

```python
import matplotlib.pyplot as plt

data = [7, 5, 8, 9, 6, 7, 5, 8, 6, 7, 6, 8, 9, 4, 7, 5, 9, 8, 6, 7]

fig = plt.figure(figsize=(6,8))  # Ukuran plot 6x8 inci

plt.violinplot(data, showmeans=True, showmedians=True)

plt.title("Violin Plot")  # Judul Plot
plt.ylabel("Nilai")  # Label sumbu Y

plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat violin plot menggunakan matplotlib untuk menampilkan distribusi data, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa list angka yang akan divisualisasikan. setelah itu kita buat figure dengan plt.figure(figsize=(6,8)) untuk mengatur ukuran tampilan plot. lalu membuat violin plot menggunakan plt.violinplot() dengan memasukkan data, serta tambahan showmeans=True untuk menampilkan rata-rata dan showmedians=True untuk menampilkan median. kemudian diberi judul menggunakan plt.title dan label sumbu y menggunakan plt.ylabel agar plot memiliki keterangan yang jelas. terakhir ditampilkan hasil plotnya dengan plt.show()

```python
import matplotlib.pyplot as plt

# Data
label = ['A', 'B', 'C', 'D', 'E']
nilai = [15, 25, 35, 20, 5]

# Membuat Plot
plt.figure(figsize=(8, 8))

# Membuat pie chart dengan .pie()
plt.pie(nilai, labels=label, autopct='%1.2f%%')

# Labeling plot
plt.title("Contoh Pie Chart")

# Menampilkan legend dan figure
plt.legend()
plt.show()
```

**Penjelasan**

program ini berfungsi untuk membuat pie chart atau diagram lingkaran menggunakan matplotlib, pertama kita import matplotlib.pyplot sebagai plt supaya bisa menggunakan fungsi plotting. kemudian kita siapkan data berupa label dan nilai, label berisi kategori seperti A sampai E, dan nilai berisi angka yang akan ditampilkan di diagram. setelah itu kita buat figure dengan plt.figure(figsize=(8, 8)) untuk mengatur ukuran tampilannya. lalu membuat pie chart menggunakan plt.pie() dengan memasukkan nilai sebagai data utama, label sebagai penanda tiap bagian, dan autopct untuk menampilkan persentase dengan 2 angka di belakang koma. kemudian diberi judul menggunakan plt.title agar diagram punya deskripsi. terakhir ditampilkan legend dan juga hasil plotnya dengan plt.legend() dan plt.show()
