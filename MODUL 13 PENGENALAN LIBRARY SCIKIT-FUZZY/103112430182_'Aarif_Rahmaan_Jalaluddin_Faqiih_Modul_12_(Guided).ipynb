## 12.2. Pendefinisian Simbol Pada SymPy

SymPy memungkinkan perhitungan matematika dilakukan secara simbolik, yang berarti bahwa variabel dan ekspresi matematika dapat didefinisikan dan dimanipulasi sebagai simbol aljabar. Dengan simbol, berbagai perhitungan dan manipulasi matematika dapat dilakukan secara simbolis. Untuk melakukan ini, langkah pertama adalah mendefinisikan simbol menggunakan `symbols()`. Berikut adalah beberapa contoh dalam penggunaan `symbols()`:

```python
# Penggunaan symbols() untuk mendefinisikan satu simbol
x = sp.symbols('x')
x
```

```python
# Pendefinisian beberapa simbol
# Cara 1
a, b, c = sp.symbols("a b c")
# Cara 2
a, b, c = sp.symbols("a, b, c")
a + b + c
```

```python
# Pendefinisian simbol dengan list
my_symbols = ['a', 'b', 'c']
a, b, c = sp.symbols(my_symbols)
a + b + c
```

```python
# Pendefinisian simbol variabel dengan indeks
x1, x2, x3 = sp.symbols("x1, x2, x3")
x1 + x2 + x3
```

```python
# Pendefinisian simbol dengan notasi range
x_n = sp.symbols('x:5')
x_n[0] + x_n[1] + x_n[2] + x_n[3] + x_n[4]
```

```python
# Pendefinisian simbol Yunani
alpha, beta, gamma = sp.symbols("alpha, beta, gamma")
alpha + beta + gamma
```

## 12.3. Aljabar Dasar Pada SymPy

SymPy memungkinkan untuk melakukan operasi aljabar dasar secara simbolik, yang berarti variabel dan ekspresi matematika dapat dimanipulasi seperti dalam aljabar. SymPy menyediakan alat untuk penjumlahan, pengurangan, perkalian, pembagian, pengembangan, dan penyederhanaan ekspresi matematika. Berikut adalah implementasi aljabar dasar pada SymPy:

```python
# Penjumlahan
x, y = sp.symbols('x, y')
addition = x + y
addition
```

```python
# Pengurangan
x, y = sp.symbols('x, y')
substraction = x - y
substraction
```

```python
# Perkalian
x, y = sp.symbols('x, y')
multiplication = x * y
multiplication
```

```python
# Pembagian
x, y = sp.symbols('x, y')
division = x / y
division
```

### `expand()`

Metode ini digunakan untuk mengembangkan ekspresi simbolik, yang menguraikan ekspresi menjadi bentuk yang lebih panjang.

```python
x, y = sp.symbols('x, y')
expr = (x + y) ** 2
expr
```

```python
# Mengembangkan ekspresi
expanded_expr = expr.expand()
expanded_expr
```

### `.simplify()`

Metode ini digunakan untuk menyederhanakan ekspresi simbolik ke bentuk yang paling sederhana. Metode ini mungkin melibatkan penggabungan suku, pemecahan faktor, penyederhanaan trigonometri, dan lain-lain untuk menemukan ekspresi yang paling sederhana.

```python
x = sp.symbols('x')
expr = sp.sin(x)**2 + sp.cos(x)**2
expr
```

```python
# Menyederhanakan ekspresi
simplified_expr = expr.simplify()
simplified_expr
```

## 12.4. Kalkulus Dasar Pada SymPy

SymPy juga dapat digunakan untuk melakukan perhitungan kalkulus dasar secara simbolik. Ini termasuk diferensiasi, integrasi, dan perhitungan limit. Dengan SymPy, operasi ini dapat dilakukan pada ekspresi matematis secara analitis, memungkinkan manipulasi dan evaluasi simbolik tanpa perlu konversi langsung ke bentuk numerik. Berikut adalah contoh implementasi kalkulus dasar pada SymPy:

### `.diff()`

Metode ini digunakan untuk melakukan diferensiasi atau menghitung turunan dari suatu fungsi simbolik terhadap variabel tertentu.

```python
x = sp.symbols('x')
# Pendefinisian fungsi f(x)
f = x**3 + 2*x**2 + x
f
```

```python
# Diferensiasi fungsi f(x) terhadap x
f_prime = f.diff(x)
f_prime
```

### `.integrate()`

Metode ini digunakan untuk melakukan integrasi dari suatu fungsi simbolik. Integrasi dapat dilakukan secara tak tentu (integral tanpa batas) atau secara tentu (integral dengan batas).

```python
x = sp.symbols('x')
# Pendefinisian fungsi f(x)
f = sp.sin(x)
f
```

```python
# Integrasi tak tentu
f_integral_indef = sp.integrate(f, x)
f_integral_indef
```

```python
# Integrasi tentu dari 0 sampai pi
f_integral_def = sp.integrate(f, (x, 0, sp.pi))
# .integrate(fungsi, (ekspresi, batas_bawah, batas_atas))
f_integral_def
```

### `.limit()`

Metode ini digunakan untuk menghitung limit dari suatu fungsi ketika variabel mendekati nilai tertentu.

```python
x = sp.symbols('x')
# Pendefinisian fungsi f(x)
f = sp.sin(x) / x
f
```

```python
# Menghitung limit f(x) saat x mendekati 0
f_limit = sp.limit(f, x, 0)
f_limit
```

## 12.5. Penyelesaian Persamaan Simbolik Pada SymPy

Sympy menyediakan alat yang kuat untuk menyelesaikan persamaan simbolik, baik persamaan aljabar maupun persamaan diferensial. Dua metode utama yang digunakan dalam SymPy untuk penyelesaian simbolik adalah `.solve()` dan `.dsolve()`. Berikut adalah implementasi dan contoh dari penyelesaian persamaan simbolik:

### `.Eq()`

Fungsi ini digunakan untuk mendefinisikan sebuah persamaan simbolik pada SymPy. Fungsi ini digunakan untuk menyatakan bahwa dua ekspresi matematika sama.

```python
x = sp.symbols('x')
# Membuat persamaan
equation = sp.Eq(x**2 - 4, 0)
equation
```

```python
x, y = sp.symbols('x, y')
# Membuat persamaan
equation = sp.Eq(x**3 - 2, y - 2 + 3*x)
equation
```

### `.solve()`

Merupakan fungsi yang digunakan untuk mencari solusi dari persamaan atau sistem persamaan simbolik. Fungsi ini memecahkan persamaan aljabar dengan mencari nilai simbolik yang memenuhi persamaan tersebut.

```python
x = sp.symbols('x')
# Membuat persamaan
equation = sp.Eq(x**2 - 4, 0)
equation
```

```python
# Memecahkan persamaan untuk x
solution = sp.solve(equation, x)
solution
```

### `.dsolve()`

Merupakan fungsi yang digunakan untuk menyelesaikan persamaan diferensial simbolik. Fungsi ini mencari solusi umum dari persamaan diferensial dalam bentuk simbolik.

```python
# Pendefinisian simbol dan fungsi
t = sp.symbols('t')
y = sp.Function('y')(t)
equation = sp.Eq(y.diff(t, t) + y, 0)
equation
```

```python
# Penyelesaian persamaan diferensial
solution = sp.dsolve(equation, y)
solution
```

## 12.6. Manipulasi Matriks Pada SymPy

SymPy menyediakan berbagai alat untuk manipulasi matriks, yang mencakup pendefinisian matriks, operasi determinan, operasi invers, dan perkalian matriks. Berikut adalah implementasi dan contoh manipulasi matriks pada SymPy:

### `.Matrix()`

Fungsi ini digunakan untuk mendefinisikan matriks yang dapat diisi dengan angka, variabel simbolik, atau ekspresi simbolik.

```python
# Pendefinisian matriks 2x2
A = sp.Matrix([[1, 2], [3, 4]])
A
```

```python
# Pendefinisian matriks nol 3x3
A = sp.zeros(3)
A
```

```python
# Pendefinisian matriks identitas 3x3
A = sp.eye(3)
A
```

### `.det()`

Fungsi ini digunakan untuk menghitung determinan dari sebuah matriks persegi.

```python
# Pendefinisian Matriks
A = sp.Matrix([[1, 2], [3, 4]])
A
```

```python
# Menghitung determinan dari matriks A
det_A = A.det()
det_A
```

### `.inv()`

Fungsi ini digunakan untuk menghitung invers dari sebuah matriks.

```python
# Pendefinisian Matriks
A = sp.Matrix([[1, 2], [3, 4]])
A
```

```python
# Menghitung invers dari matriks A
inv_A = A.inv()
inv_A
```

### Matrix Multiplication

Perkalian matriks dilakukan dengan `*`, menghasilkan matriks baru yang merupakan hasil dari perkalian dua matriks.

```python
# Pendefinisian Matriks
A = sp.Matrix([[1, 2], [3, 4]])
B = sp.Matrix([[5, 6], [7, 8]])
# Perkalian matriks A dan B
mat_mul = A * B
mat_mul
```
