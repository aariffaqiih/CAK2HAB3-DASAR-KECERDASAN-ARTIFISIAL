**f. (5 poin)** Buat persamaan berikut dengan memanfaatkan `.eq()`:

##### $ -z^2 + e^y + sin(x)= \sqrt{a^2 + b^2} + \log(z + 1)$

Tip :
Manfaatkan `sp.sin()`, `sp.exp`, `sp.sqrt`, dan `sp.log`

```python
# Definisikan simbol
x, y, z, a, b = sp.symbols('x y z a b')

# Buat persamaan
lhs = -z**2 + sp.exp(y) + sp.sin(x)  # Persamaan ruas kiri
rhs = sp.sqrt(a**2 + b**2) + sp.log(z + 1)  # Persamaan ruas kanan
pers = sp.Eq(lhs, rhs)

# Cetak persamaan
pers
```

### **2. Pengenalan *Library* Scikit-Fuzzy (60)**

Anda diminta untuk membuat **Sistem Fuzzy** yang memodelkan pemberian `Bonus` karyawan berdasarkan dua faktor utama:

- **Kinerja Karyawan** (`Performance`) : Seberapa baik karyawan bekerja, dinilai pada skala **0 hingga 100**.
- **Kehadiran** (`Attendance`) : Presentase kehadiran karyawan selama periode tertentu, juga dinilai pada skala **0 hingga 100**.

Berdasarkan penilaian ini, sistem fuzzy akan memberikan rekomendasi `Bonus` dalam rentang nilai **0 hingga 20**. Berikut informasi detailnya:

### **Input (Antecedents)**

#### Performance

- Rentang nilai: $[1, 100]$

- Fuzzy Set :
    - **Poor**: Triangular dengan titik $[0, 0, 50]$.
    - **Average**: Triangular dengan titik $[0, 50, 100]$.
    - **Excellent**: Triangular dengan titik $[50, 100, 100]$.

#### Attendance

- Rentang nilai: $[1, 100]$

- Fuzzy Set :
    - **Low**: Triangular dengan titik $[0, 0, 50]$.
    - **Medium**: Triangular dengan titik $[0, 50, 80]$.
    - **High**: Triangular dengan titik $[70, 100, 100]$.

### **Output (Consequent)**

#### Bonus

- Rentang nilai: $[0, 20]$

- Fuzzy Set :
    - **Small**: Triangular dengan titik $[0, 0, 10]$
    - **Moderate**: Triangular dengan titik $[5, 12, 18]$
    - **Large**: Triangular dengan titik $[10, 20, 20]$

### **Aturan Fuzzy**

1. Jika **Performance** `Excellent` dan **Attendance** `High`, maka **Bonus** `Large`.
2. Jika **Performance** `Average`, maka **Bonus** `Moderate`.
3. Jika **performance** `Poor` atau **Attendance** `Low`, maka **Bonus** `Small`.

Tampilkan **Hasil dan Visualisasi** dari perihtungan bonus jika
- Performance = 70
- Attendance = 85
            

**a. (5 poin)** Inisialisasi variabel input

```python
# Inisialisasi variabel input untuk performance
performance = ctrl.Antecedent(np.arange(0, 101, 1), 'performance')
# Inisialisasi variabel input untuk attendance
attendance = ctrl.Antecedent(np.arange(0, 101, 1), 'attendance')
```

**b. (5 poin)** Inisialisasi variabel output

```python
# Inisialisasi variabel output untuk bonus
bonus = ctrl.Consequent(np.arange(0, 21, 1), 'bonus')
```

**c. (10 poin)** Buat fungsi keanggotaan untuk `Performance` dan tampilkan graf fungsi keanggotaannya.

```python
# Fungsi keanggotaan untuk performance
performance['poor'] = fuzz.trimf(performance.universe, [0, 0, 50])
performance['average'] = fuzz.trimf(performance.universe, [0, 50, 100])
performance['excellent'] = fuzz.trimf(performance.universe, [50, 100, 100])

# Tampilkan fungsi keanggotaan performance
performance.view()
plt.title("Graf fungsi keanggotaan Performance")
```

Contoh *output*:

![Graf fungsi keanggotaan](./asset/1_kenaggotaan_performance.png)

**d. (10 poin)** Buat fungsi keanggotaan untuk `Attendance` dan tampilkan graf fungsi keanggotaannya.

```python
# Fungsi keanggotaan untuk attendance
attendance['low'] = fuzz.trimf(attendance.universe, [0, 0, 50])
attendance['medium'] = fuzz.trimf(attendance.universe, [0, 50, 80])
attendance['high'] = fuzz.trimf(attendance.universe, [70, 100, 100])

# Tampilkan fungsi keanggotaan attendance
attendance.view()
plt.title("Graf fungsi keanggotaan Attendance")
```

Contoh *output*:

![Graf keanggotaan attendance](./asset/2_keanggotaan_attendance.png)

**e. (10 poin)** Buat fungsi keanggotaan untuk `Bonus` dan tampilkan graf fungsi keanggotaannya.

```python
# Fungsi keanggotaan untuk bonus
bonus['small'] = fuzz.trimf(bonus.universe, [0, 0, 10])
bonus['moderate'] = fuzz.trimf(bonus.universe, [5, 12, 18])
bonus['large'] = fuzz.trimf(bonus.universe, [10, 20, 20])

# Tampilkan fungsi keanggotaan bonus
bonus.view()
plt.title("Graf fungsi keanggotaan Bonus")
```

Contoh *output* :

![Fungsi keanggotaan Bonus](./asset/4_keanggotaan_bonus.png)

**f. (10 poin)** Buat aturan fuzzy berdasarkan aturan yang sudah ditentukan

```python
# Aturan Fuzzy
rule1 = ctrl.Rule(performance['excellent'] & attendance['high'], bonus['large'])
rule2 = ctrl.Rule(performance['average'], bonus['moderate'])
rule3 = ctrl.Rule(performance['poor'] | attendance['low'], bonus['small'])
```

**g. (5 poin)** Buat sistem kontrol berdasarkan aturan yang dibuat untuk melakukan defuzzification

```python
# Membuat sistem kontrol fuzzy
bonus_ctrl = ctrl.ControlSystem([rule1, rule2, rule3])
bonus_simulation = ctrl.ControlSystemSimulation(bonus_ctrl)
```

**h. (2 poin)** Berikan input pada variabel Antecedent dengan nilai:
- Performance = 70
- Attendance = 85

```python
# Memberikan input pada variabel Antecedent
bonus_simulation.input['performance'] = 70  # Input performance
bonus_simulation.input['attendance'] = 85   # Input attendance
```

**i. (1 poin)** Hitung output dari sistem fuzzy berdasarkan input yang sudah diberikan

```python
# Menghitung output dari sistem fuzzy
bonus_simulation.compute()
```

**j. (2 poin)** Tampilkan hasil dan visualisasi dari perhitungan sistem fuzzy

```python
# Menampilkan hasil & visualisasi perhitungan
print(f"Bonus: {bonus_simulation.output['bonus']:.2f}%")
bonus.view(sim=bonus_simulation)
plt.title("Visualisasi Output Sistem Fuzzy Bonus")
```

Contoh *output*:

Bonus: 12.91%

4. Fungsi `trimf` menerima dua parameter, yaitu universe (rentang nilai dari variabel) dan tiga titik koordinat. Misalnya, `fuzz.trimf(performance.universe, [0, 0, 50])` untuk kategori 'poor' berarti fungsi keanggotaan ini berbentuk segitiga dengan titik awal di 0, titik puncak di 0, dan titik akhir di 50, sehingga semakin mendekati 0 maka derajat keanggotaannya semakin tinggi, dan semakin mendekati 50 semakin rendah.

7. Misalnya, `ctrl.Rule(performance['excellent'] & attendance['high'], bonus['large'])` berarti jika performance bernilai excellent DAN attendance bernilai high, maka bonus yang dihasilkan adalah large.

8. Aturan kedua, `ctrl.Rule(performance['average'], bonus['moderate'])`, cukup melihat performance saja, jika bernilai average maka bonus moderate.

9. Aturan ketiga menggunakan operator OR, yaitu `ctrl.Rule(performance['poor'] | attendance['low'], bonus['small'])`, yang berarti jika performance bernilai poor ATAU attendance bernilai low, maka bonus yang dihasilkan adalah small. Ketiga aturan ini kemudian dikumpulkan ke dalam satu menggunakan `ctrl.ControlSystem([rule1, rule2, rule3])`, yang berfungsi sebagai wadah untuk menyimpan seluruh aturan fuzzy.

10. Selanjutnya, `ctrl.ControlSystemSimulation(bonus_ctrl)` berfungsi membuat simulasi dari sistem kontrol yang sudah dibuat, sehingga sistem bisa menerima input numerik dan menghitung output. Input diberikan dengan cara `bonus_simulation.input['performance'] = 70` dan `bonus_simulation.input['attendance'] = 85`, yang artinya memasukkan nilai performance sebesar 70 dan attendance sebesar 85.

11. Fungsi `bonus_simulation.compute()` kemudian berfungsi menjalankan seluruh proses perhitungan fuzzy, yaitu melakukan fuzzifikasi (mengubah nilai crisp menjadi derajat keanggotaan), evaluasi aturan (menghitung mana aturan yang aktif dan seberapa besar), dan defuzzifikasi (mengubah hasil fuzzy kembali menjadi satu angka crisp).
