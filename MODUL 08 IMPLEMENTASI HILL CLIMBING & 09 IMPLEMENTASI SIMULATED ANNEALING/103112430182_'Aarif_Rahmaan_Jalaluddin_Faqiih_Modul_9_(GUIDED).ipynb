```python
import random
import math

# Fungsi untuk menghitung jarak Euclidean antara dua titik
def distance(city1, city2):
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2)

# Fungsi untuk menjumlah seluruh jarak antar titik pada rute tour
def total_distance(tour, cities):
    dist = 0
    for i in range(len(tour)):
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]])
    return dist

# Fungsi untuk melakukan random swap dua titik dalam rute
def random_swap(tour):
    new_tour = tour[:]
    i, j = random.sample(range(len(tour)), 2)
    new_tour[i], new_tour[j] = new_tour[j], new_tour[i]
    return new_tour

# Algoritma Simulated Annealing
def simulated_annealing(cities, initial_temp, cooling_rate, stopping_temp):
  # Inisialisasi solusi awal
    current_solution = list(range(len(cities))) # Rute awal (urut)
    random.shuffle(current_solution) # Mengacak rute awal
    # Menghitung total jarak untuk rute awal
    current_distance = total_distance(current_solution, cities)
    # Set best solution dan best distance berdasarkan current solution
    best_solution = current_solution[:]
    best_distance = current_distance
    # Set suhu awal berdasarkan parameter yang diterima oleh function
    temp = initial_temp

    # Program akan terus melakukan loop hingga suhu kurang dari suhu stopping
    while temp > stopping_temp:
        # Buat solusi baru dengan menukar dua titik acak
        new_solution = random_swap(current_solution)
        new_distance = total_distance(new_solution, cities)

        # Jika solusi baru lebih baik, terima solusi tersebut
        if new_distance < current_distance:
            current_solution = new_solution[:]
            current_distance = new_distance
        else:
          # jika solusi baru lebih buruk, terima dengan probabilitas tertentu (Metropolis Criterion)
            if random.random() < math.exp((current_distance - new_distance) / temp):
                current_solution = new_solution[:]
                current_distance = new_distance

        # Simpan solusi terbaik yang ditemukan
        if current_distance < best_distance:
            best_solution = current_solution[:]
            best_distance = current_distance

        # Kurangi suhu sesuai cooling rate
        temp *= cooling_rate

    return best_solution, best_distance

# Deklarasi titik-titik kota (x, y)
cities = [
    (0, 0), (1, 3), (4, 3), (6, 1),
    (3, 0), (2, 2), (5, 5), (7, 7)
]

# Parameter algoritma
initial_temp = 10000 # Suhu awal
cooling_rate = 0.995 # Laju pendinginan
stopping_temp = 1 # Suhu ketika berhenti

# Menjalankan algoritma Simulated Annealing
best_solution, best_distance = simulated_annealing(cities, initial_temp, cooling_rate, stopping_temp)

# Output hasil
print("Solusi terbaik (urutan kota):", best_solution)
print("Total jarak terbaik:", best_distance)
```

Bagian utama adalah fungsi simulated_annealing yang menerima parameter: daftar kota, suhu awal, laju pendinginan, dan suhu berhenti. Proses dimulai dengan membuat rute acak dari semua kota, lalu menghitung total jaraknya. Program menyimpan solusi terbaik yang ditemukan selama proses berjalan.

Selama suhu masih di atas nilai stopping, program terus mencoba solusi baru dengan menukar dua titik. Jika solusi baru lebih baik (jarak lebih pendek), langsung dipakai. Jika lebih buruk, masih ada kemungkinan diterima berdasarkan probabilitas yang dihitung dari selisih jarak dan suhu saat ini - ini memungkinkan program keluar dari jebakan lokal. Setiap kali ditemukan solusi lebih baik, disimpan sebagai best_solution. Suhu dikurangi terus sesuai cooling_rate sampai mencapai stopping_temp.

Di bagian bawah, delapan koordinat kota dideklarasikan sebagai tuple (x, y). Parameter algoritma diset dengan suhu awal 10000, laju pendinginan 0.995, dan suhu berhenti 1. Setelah algoritma selesai, output menampilkan urutan kota terbaik dan total jaraknya.

```python
import random
import math

# Fungsi jarak Euclidean
def distance(city1, city2):
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2)

# Total jarak rute
def total_distance(tour, cities):
    dist = 0
    for i in range(len(tour)):
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]])
    return dist

# Random swap
def random_swap(tour):
    new_tour = tour[:]
    i, j = random.sample(range(len(tour)), 2)
    new_tour[i], new_tour[j] = new_tour[j], new_tour[i]
    return new_tour

# Simulated Annealing (versi dengan alur jelas)
def simulated_annealing(cities, initial_temp, cooling_rate, stopping_temp):

    current_solution = list(range(len(cities)))
    random.shuffle(current_solution)

    current_distance = total_distance(current_solution, cities)

    best_solution = current_solution[:]
    best_distance = current_distance

    temp = initial_temp
    iteration = 1

    print("=== MULAI SIMULATED ANNEALING ===")
    print("Solusi awal :", current_solution)
    print("Jarak awal  :", current_distance)
    print()

    while temp > stopping_temp:
        print(f"--- Iterasi {iteration} ---")
        print(f"Suhu saat ini: {temp:.2f}")

        new_solution = random_swap(current_solution)
        new_distance = total_distance(new_solution, cities)

        print("Solusi sekarang :", current_solution)
        print("Jarak sekarang  :", current_distance)
        print("Solusi baru     :", new_solution)
        print("Jarak baru      :", new_distance)

        # Jika lebih baik
        if new_distance < current_distance:
            print("✔ Diterima (lebih baik)")
            current_solution = new_solution[:]
            current_distance = new_distance
        else:
            # Probabilitas penerimaan solusi buruk
            prob = math.exp((current_distance - new_distance) / temp)
            rand_val = random.random()

            print(f"Probabilitas terima: {prob:.5f}")
            print(f"Nilai random       : {rand_val:.5f}")

            if rand_val < prob:
                print("✔ Diterima (walau lebih buruk)")
                current_solution = new_solution[:]
                current_distance = new_distance
            else:
                print("✖ Ditolak")

        # Update solusi terbaik
        if current_distance < best_distance:
            best_solution = current_solution[:]
            best_distance = current_distance
            print("🔥 Update BEST solution!")

        print()

        # Turunkan suhu
        temp *= cooling_rate
        iteration += 1

    print("=== SELESAI ===\n")
    return best_solution, best_distance


# Data kota
cities = [
    (0, 0), (1, 3), (4, 3), (6, 1),
    (3, 0), (2, 2), (5, 5), (7, 7)
]

# Parameter
initial_temp = 10000
cooling_rate = 0.995
stopping_temp = 1

# Jalankan
best_solution, best_distance = simulated_annealing(
    cities, initial_temp, cooling_rate, stopping_temp
)

print("=== HASIL AKHIR ===")
print("Solusi terbaik:", best_solution)
print("Total jarak   :", best_distance)
```

```python
import random
import math

# Hitung jarak Euclidean antar dua kota
def distance(city1, city2):
    return math.hypot(city1[0] - city2[0], city1[1] - city2[1])

# Hitung total jarak rute salesman
def total_distance(tour, cities):
    dist = 0
    for i in range(len(tour)):
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]])
    return dist

# Bagi solusi ke dalam dua rute
def split_solution(solution):
    half = len(solution) // 2
    return solution[:half], solution[half:]

# Lakukan pertukaran dua elemen secara acak
def random_swap(solution):
    new_solution = solution[:]
    i, j = random.sample(range(len(solution)), 2)
    new_solution[i], new_solution[j] = new_solution[j], new_solution[i]
    return new_solution

# Simulated Annealing untuk mTSP dua salesman
def simulated_annealing_mTSP(cities, initial_temp, cooling_rate, stopping_temp):
    n = len(cities)
    current_solution = list(range(n))
    random.shuffle(current_solution)
    current_r1, current_r2 = split_solution(current_solution)
    current_distance = total_distance(current_r1, cities) + total_distance(current_r2, cities)

    best_solution = current_solution[:]
    best_distance = current_distance

    temp = initial_temp
    while temp > stopping_temp:
        new_solution = random_swap(current_solution)
        new_r1, new_r2 = split_solution(new_solution)
        new_distance = total_distance(new_r1, cities) + total_distance(new_r2, cities)

        if new_distance < current_distance or random.random() < math.exp((current_distance - new_distance) / temp):
            current_solution = new_solution
            current_distance = new_distance
            if current_distance < best_distance:
                best_solution = new_solution[:]
                best_distance = current_distance

        temp *= cooling_rate

    best_r1, best_r2 = split_solution(best_solution)
    return best_r1, best_r2, best_distance

# Koordinat kota dari gambar
cities = [
    (10, 20), # 0
    (35, 15), # 1
    (3, 40),  # 2
    (25, 30), # 3
    (15, 25), # 4
    (40, 35), # 5
    (20, 40), # 6
    (50, 10)  # 7
]

# Parameter SA
initial_temp = 10000
cooling_rate = 0.995
stopping_temp = 1

# Eksekusi
best_r1, best_r2, best_total_distance = simulated_annealing_mTSP(cities, initial_temp, cooling_rate, stopping_temp)

# Output
print("Rute Salesman 1:", best_r1)
print("Rute Salesman 2:", best_r2)
print("Total Jarak Gabungan Terbaik:", best_total_distance)
```
