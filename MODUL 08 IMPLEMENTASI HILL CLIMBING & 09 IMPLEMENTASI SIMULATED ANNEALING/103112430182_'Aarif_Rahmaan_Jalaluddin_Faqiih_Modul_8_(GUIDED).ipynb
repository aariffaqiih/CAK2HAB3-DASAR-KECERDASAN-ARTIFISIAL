```python
# Import library yang digunakan
import random
import math

# Menghitung jarak 2 titik dengan rumus Euclidean
def distance(city1, city2):
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2)

# Menghitung total jarak dari rute tour
def total_distance(tour, cities):
    dist = 0
    for i in range(len(tour)):
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]])
    return dist

# Menghasilkan semua tetangga dari tour dengan menukar 2 titik
def get_neighbors(tour):
    neighbors = []
    for i in range(len(tour)):
        for j in range(i + 1, len(tour)):
            neighbor = tour[:]
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append(neighbor)
    return neighbors

# Deklarasi titik-titik kota dalam bentuk koordinat (x, y)
cities = [
    (0, 0), (1, 3), (4, 3), (6, 1),
    (3, 0), (2, 2), (5, 5), (7, 7)
]

# Membuat rute awal secara acak
current_solution = list(range(len(cities)))
random.shuffle(current_solution)

# Menghitung total jarak dari rute awal
current_distance = total_distance(current_solution, cities)

# Algoritma Hill Climbing
while True:
    # Mendapatkan seluruh tetangga dari solusi saat ini
    neighbors = get_neighbors(current_solution)

    # Mencari tetangga dengan total jarak terkecil
    best_neighbor = min(neighbors, key=lambda tour: total_distance(tour, cities))
    best_distance = total_distance(best_neighbor, cities)

    # Jika tidak ada solusi yang lebih baik, keluar dari loop
    if best_distance >= current_distance:
        break

    # Update solusi saat ini
    current_solution = best_neighbor
    current_distance = best_distance

# Menampilkan hasil akhir
print("Solusi terbaik (urutan kota):", current_solution)
print("Total jarak terbaik:", current_distance)
```

Jika jarak tetangga itu lebih kecil dari jarak rute saat ini, maka rute saat ini diganti dengan tetangga itu. Jika tidak ada tetangga yang lebih baik, proses berhenti.
```python
import random
import math

# Menghitung jarak 2 titik dengan rumus Euclidean
def distance(city1, city2):
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2)

# Menghitung total jarak dari rute tour
def total_distance(tour, cities):
    dist = 0
    for i in range(len(tour)):
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]])
    return dist

# Menghasilkan semua tetangga dari tour dengan menukar 2 titik
def get_neighbors(tour):
    neighbors = []
    for i in range(len(tour)):
        for j in range(i + 1, len(tour)):
            neighbor = tour[:]
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append(neighbor)
    return neighbors

# Deklarasi titik-titik kota
cities = [
    (0, 0), (1, 3), (4, 3), (6, 1),
    (3, 0), (2, 2), (5, 5), (7, 7)
]

# Membuat rute awal secara acak
current_solution = list(range(len(cities)))
random.shuffle(current_solution)

current_distance = total_distance(current_solution, cities)

print("=== MULAI HILL CLIMBING ===")
print("Solusi awal:", current_solution)
print("Jarak awal:", current_distance)
print()

iteration = 1

while True:
    print(f"--- Iterasi {iteration} ---")

    neighbors = get_neighbors(current_solution)
    print(f"Jumlah tetangga: {len(neighbors)}")

    # Cari tetangga terbaik
    best_neighbor = min(neighbors, key=lambda tour: total_distance(tour, cities))
    best_distance = total_distance(best_neighbor, cities)

    print("Solusi saat ini :", current_solution)
    print("Jarak saat ini  :", current_distance)
    print("Tetangga terbaik:", best_neighbor)
    print("Jarak tetangga  :", best_distance)

    # Cek apakah lebih baik
    if best_distance >= current_distance:
        print("Tidak ada perbaikan, STOP.")
        break

    print("Solusi diperbarui!")
    print()

    current_solution = best_neighbor
    current_distance = best_distance

    iteration += 1

print("\n=== HASIL AKHIR ===")
print("Solusi terbaik:", current_solution)
print("Total jarak   :", current_distance)
```

Kode ini berfungsi untuk menemukan rute terpendek dalam masalah traveling salesman menggunakan algoritma hill climbing. Terdapat fungsi distance untuk menghitung jarak antara dua titik koordinat euclidean, sedangkan total_distance menghitung panjang seluruh rute perjalanan. Fungsi get_neighbors menciptakan variasi rute baru dengan cara menukar posisi dua kota. Kumpulan koordinat kota didefinisikan dalam daftar cities dan rute awal dibuat secara acak. Proses utama berjalan pada loop tak terbatas yang mengevaluasi setiap tetangga untuk mencari rute dengan jarak paling kecil. Jika rute terbaik tersebut lebih pendek dari rute saat ini, maka rute tersebut diadopsi sebagai solusi baru. Iterasi akan berhenti ketika tidak ditemukan tetangga yang memberikan jarak lebih pendek lagi.

```python
import random

# Tabel jarak antar kota (dictionary)
distance_matrix = {
    'A': {'A': 0, 'B': 10, 'C': 15, 'D': 20, 'E': 25},
    'B': {'A': 10, 'B': 0, 'C': 35, 'D': 25, 'E': 15},
    'C': {'A': 15, 'B': 35, 'C': 0, 'D': 30, 'E': 20},
    'D': {'A': 20, 'B': 25, 'C': 30, 'D': 0, 'E': 10},
    'E': {'A': 25, 'B': 15, 'C': 20, 'D': 10, 'E': 0}
}

# Fungsi menghitung total jarak rute
def total_distance(tour, distance_matrix):
    dist = 0
    for i in range(len(tour) - 1):  # Jangan hitung kota terakhir-ke-awal secara manual
        dist += distance_matrix[tour[i]][tour[i + 1]]
    dist += distance_matrix[tour[-1]][tour[0]]  # Tambahkan kembali ke kota awal
    return dist

# Fungsi menghasilkan semua neighbors
def get_neighbors(tour):
    neighbors = []
    # Tukar dua kota di rute (kecuali kota awal dan akhir)
    for i in range(1, len(tour) - 1):
        for j in range(i + 1, len(tour) - 1):
            neighbor = tour[:]
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append(neighbor)
    return neighbors

# Inisialisasi solusi awal
cities = ['A', 'B', 'C', 'D', 'E']
current_solution = cities[:]
random.shuffle(current_solution[1:])  # Acak kota selain kota pertama
current_solution.append('A')  # Tambahkan kembali ke kota awal

# Output solusi awal
print("Solusi awal:", " → ".join(current_solution))
current_distance = total_distance(current_solution, distance_matrix)

# Hill Climbing
while True:
    # Hasilkan semua neighbors
    neighbors = get_neighbors(current_solution)

    # Cari neighbor dengan jarak terbaik
    best_neighbor = min(neighbors, key=lambda tour: total_distance(tour, distance_matrix))
    best_distance = total_distance(best_neighbor, distance_matrix)

    # Jika tidak ada perbaikan, keluar dari loop
    if best_distance >= current_distance:
        break

    # Perbarui solusi saat ini
    current_solution = best_neighbor
    current_distance = best_distance

# Output hasil akhir
print("Solusi terbaik (urutan kota):", " → ".join(current_solution))
print("Total jarak terbaik:", current_distance)
```
