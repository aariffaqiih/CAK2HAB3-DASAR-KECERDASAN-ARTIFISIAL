# Guided 1 (Implementasi BFS pada Tree)

```python
# Import library
import networkx as nx  # untuk membuat grafik
import matplotlib.pyplot as plt  # untuk plot grafik


# Fungsi pendukung untuk mencetak graf
def show_graph(G, pos=None, title=''):
    # Membuat pos jika pos tidak diberikan
    if pos is None:
        pos = nx.spring_layout(G)

    # Fungsi untuk menggambar node
    nx.draw(
        G,  # Graf NetworkX
        pos,  # Posisi node
        with_labels=True,  # Menampilkan nama node
        node_color='red',  # Warna node
        node_size=2000,  # Ukuran node
        font_color="white",  # Warna font label node
        font_weight="bold",  # Ketebalan font label node
        width=5  # Ketebalan garis edge
    )

    # Mengambil label edge jika ada weight
    edge_labels = nx.get_edge_attributes(G, 'weight')

    # Fungsi untuk menggambar label edge
    nx.draw_networkx_edge_labels(
        G,
        pos,
        edge_labels=edge_labels,  # Data weight
        font_color='blue',  # Warna font label edge
        font_weight="bold",  # Berat font label edge
        font_size=12,  # Ukuran font label edge
    )

    plt.margins(0.2)  # Memberikan margin pada plot
    plt.title(title)  # Menampilkan judul graf jika diberikan
    plt.show()  # Menampilkan graf menggunakan matplotlib


# Posisi node
pos = {
    'A': (0, 3),
    'B': (-10, 2), 'C': (10, 2),
    'D': (-15, 1), 'E': (-5, 1), 'F': (5, 1), 'G': (15, 1),
    'H': (-18, 0), 'I': (-12, 0), 'J': (-8, 0), 'K': (-2, 0),
    'L': (2, 0), 'M': (8, 0), 'N': (12, 0), 'O': (18, 0)
}


# Inisialisasi graf kosong
G = nx.Graph()

# Definisi node dan edge
edges = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'F'), ('C', 'G'),
    ('D', 'H'), ('D', 'I'),
    ('E', 'J'), ('E', 'K'),
    ('F', 'L'), ('F', 'M'),
    ('G', 'N'), ('G', 'O')
]

# Menambahkan edge ke graf
G.add_edges_from(edges)


# Evaluasi elemen pada graf
print("Elemen pada Graf:")
print("Daftar Node pada Graf:", G.nodes())
print("Daftar Edge pada Graf:", G.edges())
print("Jumlah Node pada Graf:", G.number_of_nodes())
print("Jumlah Edge pada Graf:", G.number_of_edges())


# Visualisasi graf
show_graph(G, pos=pos)
```

```python
# Urutan Breadth-First Search
print("Urutan node yang dikunjungi:")
i = 1

for nodes in nx.bfs_edges(G, source='A'):
    print(f"{i}. {nodes[0]} - {nodes[1]}")
    i += 1


# BFS dengan depth limit = 2
print("Urutan node yang dikunjungi:")
i = 1

for nodes in nx.bfs_edges(G, source='A', depth_limit=2):
    print(f"{i}. {nodes[0]} - {nodes[1]}")
    i += 1
```

# Guided 4 (Implementasi BFS pada Directed Graph)

```python
# Import library
import networkx as nx  # untuk membuat grafik
import matplotlib.pyplot as plt  # untuk plot grafik


# Fungsi pendukung untuk mencetak graf
def show_graph(G, pos=None, title='', labels=None):
    # Membuat pos jika pos tidak diberikan
    if pos is None:
        pos = nx.spring_layout(G)

    # Fungsi untuk menggambar node
    nx.draw(
        G,  # Graf NetworkX
        pos,  # Posisi node
        with_labels=True,  # Menampilkan nama node
        labels=labels,  # Menampilkan label setiap node
        node_color='red',  # Warna node
        node_size=2000,  # Ukuran node
        font_color="white",  # Warna font label node
        font_weight="bold",  # Ketebalan font label node
        width=5  # Ketebalan garis edge
    )

    # Mengambil label edge jika ada weight
    edge_labels = nx.get_edge_attributes(G, 'weight')

    # Fungsi untuk menggambar label edge
    nx.draw_networkx_edge_labels(
        G,
        pos,
        edge_labels=edge_labels,  # Data weight
        font_color='blue',  # Warna font label edge
        font_weight="bold",
        font_size=12,
    )

    plt.margins(0.2)
    plt.title(title)
    plt.show()


# Inisialisasi graf berarah kosong
G = nx.DiGraph()

# Definisi node dan edge
edges = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'E'), ('C', 'F'),
    ('E', 'D'), ('E', 'F')
]

# Menambahkan edge pada graf
G.add_edges_from(edges)


# Posisi node
pos = {
    'A': (0, 1),
    'B': (1, 2),
    'C': (1, 0),
    'D': (2, 3),
    'E': (2, 1.5),
    'F': (2, 0)
}


# Menambahkan label start dan goal
labels = {node: node for node in G.nodes()}
labels['A'] = 'A\nStart'
labels['F'] = 'F\nGoal'


# Evaluasi elemen pada graf
print("Elemen pada Graf:")
print("Daftar Node pada Graf:", list(G.nodes()))
print("Daftar Edge pada Graf:", list(G.edges()))
print("Jumlah Node pada Graf:", G.number_of_nodes())
print("Jumlah Edge pada Graf:", G.number_of_edges())


# Visualisasi graf
show_graph(G, pos=pos, labels=labels)
```

# Unguided Eropa

```python
import networkx as nx

# Inisialisasi graf
G_eropa = nx.Graph()

# Tambahkan edge dengan bobot
edges_eropa = [
    ("Arad", "Zerind", 75), ("Arad", "Timisoara", 118), ("Arad", "Sibiu", 140),
    ("Zerind", "Oradea", 71), ("Oradea", "Sibiu", 151),
    ("Timisoara", "Lugoj", 111), ("Lugoj", "Mehadia", 70),
    ("Sibiu", "Fagaras", 99), ("Sibiu", "Rimnicu Vilcea", 80),
    ("Rimnicu Vilcea", "Pitesti", 97)
]
G_eropa.add_weighted_edges_from(edges_eropa)

# Jalankan BFS dari Arad
print("=== BFS Eropa (Dari Arad) ===")
print("Urutan kunjungan edge:")
for i, (u, v) in enumerate(nx.bfs_edges(G_eropa, source="Arad"), 1):
    print(f"{i}. {u} -> {v} (Jarak: {G_eropa[u][v]['weight']} km)")
print("\nApakah BFS mempertimbangkan bobot edge?")
print("Tidak, BFS hanya melihat keterhubungan node dan tidak mempertimbangkan bobot.")

# Output node saja (opsional)
print("\nUrutan kunjungan node:")
print(list(nx.bfs_tree(G_eropa, source="Arad")))
```
