# **Penjelasan**

### 1. Pendefinisian Simbol

Di SymPy, setiap variabel matematika harus didefinisikan sebagai simbol terlebih dahulu.  
Contoh: `x = sp.symbols('x')`  
Ini membuat `x` sebagai variabel aljabar. Untuk beberapa simbol sekaligus, `sp.symbols('x y')` atau `sp.symbols('x:5')`.

---

### 2. Membuat Persamaan

Fungsi `sp.Eq(kiri, kanan)` menghasilkan persamaan.  
Contoh: `sp.Eq(x - 2, 0)` mewakili `x - 2 = 0`.  
Tanpa `Eq`, SymPy hanya akan melihat ekspresi `x-2`, bukan persamaan.  
`sp.Eq` digunakan sebelum mencari akar dengan `sp.solve`.

---

### 3. Menyelesaikan Persamaan Aljabar

`sp.solve(equation, x)` mengembalikan akar dalam bentuk list.

---

### 4. Turunan (Diferensiasi)

Metode `.diff(x)` pada ekspresi menghasilkan turunan terhadap `x`.  
Misal `f = sp.sin(7*x)` lalu `f.diff(x)` menjadi `7*cos(7*x)`.  
Notasi `dy/dx` di python menjadi `f.diff(x)`.  
Hasil turunan dibungkus dengan `sp.Eq(sp.Derivative(y, x), dy_dx)` untuk menampilkan bentuk persamaan diferensial, walaupun yang dihitung sebenarnya turunan biasa.

---

### 5. Integral

Integral tak tentu menggunakan `sp.integrate(f, x)`.  
Integral tentu dengan `sp.integrate(f, (x, batas_bawah, batas_atas))`.  
  
`sp.Rational(1,3)` digunakan untuk mencegah konversi ke float `0.333...`.  
`sp.Integral(f, (w,1,8))` hanya menampilkan notasi integral tanpa menghitung. Hasil hitungan diperoleh dengan `sp.integrate(...)`, lalu ditampilkan sebagai `sp.Eq(integral_expr, result)`.

---

### 6. Matriks dan Determinan

Matriks didefinisikan dengan `sp.Matrix([ [baris1], [baris2], ...])`.  
  
Determinan dihitung dengan `.det()` menjadi `-6`.

---

### 7. Penanganan Bilangan Rasional dan Simbol Yunani

Misalnya pada integral `x**(4/3)`. Agar tidak menjadi float, maka digunakan `x**sp.Rational(4,3)`.  
Ini mengubah dari pecahan eksak ke objek `Rational`.  

Simbol yunani α, β, γ dapat langsung didefinisikan dengan `sp.symbols('alpha beta gamma')`.  
SymPy secara otomatis menampilkan `α β γ`.

---

### 8. Menampilkan Ekspresi dalam Bentuk Matematis

Notebook menggunakan `Equation` dan `Integral` tanpa agar tampilannya sama seperti tulis tangan.  
Fungsi `sp.Eq(dy_dx, hasil)` digunakan untuk menuliskan `dy/dx = ...`.
